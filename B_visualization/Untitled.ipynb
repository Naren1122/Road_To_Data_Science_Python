```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
employees = ['Bob','John','Hari','Shyam','Goru']
salary = [45000,20000,43000,47000,48000]
expense = [30000,35000,22025,23000,10000]
```

```python
plt.figure(figsize = (6,4)) # 600 x 400
plt.plot(employees,salary,color = 'red', marker='o' ,linestyle = '-.')
plt.xlabel('Employees')
plt.ylabel('Salary')
plt.title('Total salary of employee')
plt.grid(linestyle = 'dotted', color = 'green', alpha =0.3)
plt.savefig('emp_salary.png')
plt.show()
```

```python
plt.figure(figsize = (6,4)) # 600 x 400
plt.plot(employees,salary,color = 'red', marker='o', linestyle = '-.',label='salary')
plt.plot(employees,expense,label='expense')
plt.xlabel('Employees')
plt.ylabel('Salary and expense')
plt.title('Total salary of employee')
plt.grid(linestyle = 'dotted', color = 'green', alpha =0.3)
plt.legend(loc = "best") # for separating the plot which plot is it and alway use label in label in plt.plot
plt.savefig('emp_salary.png')
plt.show()
```

# Bar Plot

```python
plt.figure(figsize=(6,4))
colors = ['red','blue','green','yellow']
plt.bar(employees, salary,color=colors,edgecolor = 'black')
plt.xlabel('Employee')
plt.ylabel('Salary')
plt.title('Total Salary of Employees')

for idx, val in enumerate(salary):
    plt.text(idx,val, str(val),ha='center',va='bottom',fontsize='x-small',fontstyle='italic', fontweight='bold')
plt.show()
```

# Piechart and Donut Chart

```python
ps_pct = hr.groupby('PerformanceScore')['EmpID'].count().reset_index()
ps_pct
```

```python
plt.pie(
    ps_pct['EmpID'],
    labels = ps_pct['PerformanceScore'],
    autopct = '%1.1f%%'
)
plt.title('Total Percentage of Employees by Performance Score')
plt.show()
```

```python
plt.pie(
    ps_pct['EmpID'],
    labels = ps_pct['PerformanceScore'],
    autopct = '%1.1f%%',

    # width egge Properties

    wedgeprops = {'width': 0.5,'edgecolor': '#e5e5e5'},
    pctdistance = 0.8
)
plt.title('Total Percentage of Employees by Performance Score')
plt.show()
```

# sub-plots

```python
dep_details = hr.groupby('Department').agg(
    Total_Employees = ('EmpID','count'),
    Total_Salary = ('Salary','sum'),
    Total_Late_Days = ('DaysLateLast30','sum'),
    Total_Absences = ('Absences','sum')
).reset_index()
dep_details
```

```python
def addBarLabel(dataframe,ax,column_name):
    for idx,val in enumerate (dataframe[column_name]):
        ax.text(idx,val,str(val),ha='center',va = 'bottom')
```

```python
fig , axes = plt.subplots(2,2,figsize=(14,10))
plt.subplots_adjust(hspace = 0.5)
plt.suptitle('Analysis of Departments with other features',fontsize= 'xx-large',fontfamily='monospace')

x_axis = dep_details['Department']
x_label = 'Department'

ax1 = axes[0,0]
ax1.bar(x_axis,dep_details['Total_Employees'])
ax1.set_xlabel(x_label)
ax1.set_ylabel('Total Employees')
ax1.set_title('Total Employees by Department')
ax1.tick_params(axis='x',rotation = 10)
addBarLabel(dep_details,ax1,'Total_Employees')

ax2 = axes[0,1]
ax2.bar(x_axis,dep_details['Total_Salary'])
ax2.set_xlabel(x_label)
ax2.set_ylabel('Total Salary')
ax2.set_title('Total Salary by Department')
ax2.tick_params(axis='x',rotation = 10)
addBarLabel(dep_details,ax2,'Total_Salary')


ax3 = axes[1,0]
ax3.plot(x_axis,dep_details['Total_Late_Days'])
ax3.set_xlabel(x_label)
ax3.set_ylabel('Total_Late_Days')
ax3.set_title('Total_Late_Days')
ax3.tick_params(axis='x',rotation = 10)
addBarLabel(dep_details,ax3,'Total_Late_Days')


ax4 = axes[1,1]
ax4.pie(dep_details['Total_Absences'],
        labels = x_axis,
        autopct = '%1.1f%%',
        wedgeprops = {'width':0.4},
        pctdistance = 0.8
       )
ax4.set_title('Total Absence Days by Department')
plt.savefig('Subplot.png')
plt.show()
```
