# NUMPY INDEXING AND SLICING OF MATRIX

```python
arr[5]
```

#ravel : change any dimension into 1 d array



```python
ravel_num = np.ravel(arr)
ravel_num
```

```python
val = arr[0,4]
print(val)
```

```python
#slicing the matrix
slice = arr[0:2 , 1:3]
slice
```
