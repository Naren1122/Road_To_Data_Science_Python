```python
# to convert object into datatime data type
# to_datatime(column_name, format = 'what type of date format does my dataset has? ', dayfirst = True/false(default)) 

# dayfirst = True/false(default)  use this is none of the format works only in the syntax and never keep it in first option.
'''
format
-----------
1. '%d.%m.%y'
2. '%d/%m/%y'
3. 'mixed'
'''
```

```python
hr['DOB'] = pd.to_datetime(hr['DOB'], format='%d.%m.%y')
```

```python
# since dates need to be in date_time dtype so to convert first so we create a function for it.
def convertDateTime(dataframe , *args):
    for col in args:
        dataframe[col] = pd.to_datetime(dataframe[col], format='mixed')
```
