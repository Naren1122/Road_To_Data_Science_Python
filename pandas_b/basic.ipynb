```python
df = pd.read_csv('Employees.csv')
```

```python
emp_salary = df[(df['SALARY'] > 100000) | (df['SALARY'] < 18000)][['FIRST_NAME', 'LAST_NAME', 'HIRE_DATE', 'SALARY']].head()
emp_salary
```

```python
emp_salary_bet = df[(df['SALARY']).between(10000,18000)]
emp_salary_bet
```

```python
# find employees details who works in job AD_VP and IT_PROG
emp_job = df[(df['JOB_ID'] == 'AD_VP') | (df['JOB_ID'] == 'IT_PROG')]
emp_job
```

```python
emp_job_in = df[df['JOB_ID'].isin(['AD_VP','IT_PROG'])]
emp_job_in
```

```python
#find employee details who earns more than 8000 and works in department 70 or 60
```

```python
emp_details = df[(df['SALARY'] > 8000) & ((df['DEPARTMENT_ID'] == 60) | (df['DEPARTMENT_ID'] == 70))]
emp_details
```

```python
emp_details = df[(df['SALARY'] > 8000) & (df['DEPARTMENT_ID'].isin([60, 70]))]
emp_details
```
