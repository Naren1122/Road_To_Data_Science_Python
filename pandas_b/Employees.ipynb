```python
# df.dropna() # this is temporory change , but for permanent change always use inplace = True

# dropna(axis = 0(default)/1 , how = 'any'(default)/'all' , inplace = True/False(default))

# how = any  -> any 1 null value , it will remove whole column.
# how  = all -> have to be all null value to get removed.



df.dropna(how = 'all', inplace =True)
df

```

```python
# commission percentage
com_mean = round(df['COMMISSION_PCT'].mean(),2)
com_median = df['COMMISSION_PCT'].median()

print (com_mean)
print (com_median)
```

```python
df['COMMISSION_PCT'].skew()
```

```python
# df['COMMISSION_PCT'].fillna(com_mean, inplace = True) OR

df.fillna({'COMMISSION_PCT' : com_mean}, inplace = True)
```
