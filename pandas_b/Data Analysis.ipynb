```python
# when using group by must use aggregate but if u use aggregate not neccessaily use aggregate.

## Aggregate Function -> KPI -> Key Performance Indicator('mean','max','min','count') -> KPI making data to be seen in summmary.

## group by (categorical) + aggregate function (Categorical, Discrete,continous) -> Analysis
'''
for continuous use  --> avg, max, min but not count
for discrete --> count,max , min , sum but not mean
'''
```

```python
emp_dep = df.groupby('Department')['EmpID'].count().reset_index().rename(
    columns = {'EmpID' : 'Total Employees'}
).sort_values(ascending = False, by = 'Total Employees' , ignore_index = True)
emp_dep
```

```python
emp_dep = df.groupby('Department')['EmpID'].count().rename('Total_Employees').sort_values(ascending =False)

format_data =  emp_dep.to_markdown(numalign ='center', stralign = 'center')
print(format_data)
```

```python
# Find total salary distribution to each employee by their job_position.
t_sal = df.groupby('Position')['Salary'].sum().reset_index().sort_values(ascending = False , by = 'Salary'
                                                                        ,ignore_index = True)
t_sal['Salary'] = t_sal['Salary'].map("Rs.{:,.2f}".format)
format_t = t_sal.to_markdown(numalign='center', stralign='center')

print(format_t)
```

```python
def currencyConverter(dataframe, prefix, *args):
    for col in args:
        dataframe[col] = dataframe[col].map(f'{prefix}'"{:,.2f}".format)
```

```python
emp_pos_details = df.groupby('Position').agg(
    Total_Salary = ('Salary', 'sum'),
    Total_Employees = ('EmpID', 'count'),
    Max_Salary = ('Salary', 'max'),
    Min_Salary = ('Salary', 'min'),
    Avg_Salary = ('Salary', 'mean')
).reset_index()

emp_pos_details.columns = emp_pos_details.columns.str.replace('_', ' ') 

currencyConverter(emp_pos_details, 'Rs.', 'Total Salary', 'Max Salary','Min Salary', 'Avg Salary')

emp_pos_details
```

```python
def convertDateTime(dataframe , *args):
    for col in args:
        dataframe[col] = pd.to_datetime(dataframe[col], format='mixed')
```

```python
convertDateTime(df, 'DOB','DateofHire','DateofTermination','LastPerformanceReview_Date')
```

```python
df['HireYear'] = df['DateofHire'].dt.year
df['HireMonth'] = df['DateofHire'].dt.month
df['HireDay']= df['DateofHire'].dt.day
```

```python
df['HireMonthName'] = df['DateofHire'].dt.month_name()
df['HireDayName'] = df['DateofHire'].dt.day_name()
```

```python
# Find total employees hired in each different year
```

```python
emp_year = df.groupby('HireYear')['EmpID'].count().reset_index().rename(
    columns={'HireYear': 'Hire Year', 'EmpID': 'Total Employees'}
)
emp_year['Hire_PCT_YOY'] = (emp_year['Total Employees'].pct_change().fillna(0)) * 100
emp_year['Hire_PCT_YOY'] = emp_year['Hire_PCT_YOY'].map('{:,.2f}%'.format)
emp_year
```

```python
month_emp = df.groupby(['HireMonth','HireMonthName'])['EmpID'].count().reset_index()[['HireMonthName','EmpID']]
month_emp
```

```python
day_dict ={
    'Sunday' :1,
    'Monday' : 2,
    'Tuesday' : 3,
    'Wednesday' : 4,
    'Thursday' : 5,
    'Friday' : 6,
    'Saturday' : 7,
}
df['day_dict'] = df['HireDayName'].map(day_dict)
```

```python
day_name = (df.groupby(['day_dict','HireDayName'])['EmpID']
            .count()
            .reset_index()
            [['HireDayName','EmpID']])
day_name
```
